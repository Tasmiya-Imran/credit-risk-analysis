```python
import pandas as pd
import numpy as np
import os
from datetime import datetime, timedelta

print("--- Step 1: Synthesizing 500,000+ Corporate Transaction Records ---")

# Set reproducible parameters for a large-scale financial simulation
np.random.seed(42)
num_records = 525000
base_date = datetime(2026, 1, 1)

# Generate realistic enterprise transactional columns
customer_ids = [f"CUST_{i:06d}" for i in np.random.randint(100000, 150000, num_records)]
transaction_amounts = np.round(np.random.exponential(scale=3000, size=num_records) + 500, 2)
credit_scores = np.random.randint(300, 850, num_records)

# Simulate realistic corporate risk distributions (Lower credit score = Higher delinquency likelihood)
delinquency_base = (850 - credit_scores) / 550
noise = np.random.uniform(0, 0.3, num_records)
is_delinquent = np.where((delinquency_base + noise) > 0.75, 1, 0)

# Generate historical time stamps spanning a business cycle
days_offset = np.random.randint(0, 260, num_records)
timestamps = [ (base_date + timedelta(days=int(d))).strftime('%Y-%m-%d') for d in days_offset ]

print("⚙️ Processing transaction matrices and parsing layout columns...")
df_warehouse = pd.DataFrame({
    'Transaction_Date': timestamps,
    'Customer_ID': customer_ids,
    'Credit_Score': credit_scores,
    'Transaction_Amount_INR': transaction_amounts,
    'Account_Balance_INR': np.round(transaction_amounts * np.random.uniform(1.5, 10, num_records), 2),
    'Delinquency_Flag': is_delinquent
})

# Inject artificial corporate missing data rows to simulate real-world ETL cleansing requirements
mask = np.random.rand(num_records) < 0.03
df_warehouse.loc[mask, 'Account_Balance_INR'] = np.nan

# Save the massive production file cleanly to your cloud instance disk
os.makedirs('tata_iq_warehouse', exist_ok=True)
output_path = 'tata_iq_warehouse/raw_customer_transactions.csv'
df_warehouse.to_csv(output_path, index=False)

print(f"\n🎉 SUCCESS: Formatted {len(df_warehouse):,} financial data rows securely stored at '{output_path}'!")
```
